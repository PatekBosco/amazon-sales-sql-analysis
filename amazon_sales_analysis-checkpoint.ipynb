```python
import sqlite3
import pandas as pd

# Подключаемся к нашей базе данных
conn = sqlite3.connect('my_project_db.db')

# Пишем любой SQL-запрос внутри тройных кавычек
query = """
SELECT * 
FROM amazon_sales 
LIMIT 5;
"""

# Запускаем и сразу выводим в красивом виде Pandas
df_result = pd.read_sql_query(query, conn)
df_result
```

```python
query_1 = """
SELECT 
    Category AS "Категория",
    SUM(Amount) AS "Общая выручка (INR)",
    COUNT("Order ID") AS "Количество заказов",
    ROUND(AVG(Amount), 2) AS "Средний чек"
FROM amazon_sales
WHERE Status NOT LIKE '%Cancelled%'
GROUP BY Category
ORDER BY "Общая выручка (INR)" DESC
LIMIT 5;
"""
pd.read_sql_query(query_1, conn)
```

```python
query_2 = """
SELECT 
    Category AS "Категория",
    COUNT("Order ID") AS "Количество отмен",
    ROUND(SUM(Amount), 2) AS "Упущенная выручка (INR)"
FROM amazon_sales
WHERE Status LIKE '%Cancelled%'
GROUP BY Category
ORDER BY "Упущенная выручка (INR)" DESC;
"""
pd.read_sql_query(query_2, conn)
```

```python
query_3 = """
SELECT 
    CASE WHEN B2B = 1 THEN 'Бизнес (B2B)' ELSE 'Розница (B2C)' END AS "Тип клиента",
    COUNT("Order ID") AS "Всего заказов",
    ROUND(SUM(Amount), 2) AS "Общая сумма покупок",
    ROUND(AVG(Amount), 2) AS "Средний чек одного заказа"
FROM amazon_sales
WHERE Status NOT LIKE '%Cancelled%'
GROUP BY B2B;
"""
pd.read_sql_query(query_3, conn)
```

```python
query_dynamics = """
SELECT 
    -- Превращаем текстовую дату 'MM-DD-YY' в правильный формат 'YYYY-MM-DD' для сортировки
    '20' || SUBSTR(Date, 7, 2) || '-' || SUBSTR(Date, 1, 2) || '-' || SUBSTR(Date, 4, 2) AS "Дата",
    COUNT("Order ID") AS "Количество заказов",
    ROUND(SUM(Amount), 2) AS "Выручка за день (INR)",
    ROUND(AVG(Amount), 2) AS "Средний чек"
FROM amazon_sales
WHERE Status NOT LIKE '%Cancelled%' -- Считаем только успешные продажи
  AND Date IS NOT NULL
GROUP BY "Дата"
ORDER BY "Дата" ASC; -- Сортируем от самых ранних к самым поздним
"""

df_dynamics = pd.read_sql_query(query_dynamics, conn)
df_dynamics
```

```python
import matplotlib.pyplot as plt

# Преобразуем колонку с датой в специальный формат для графиков
df_dynamics['Дата'] = pd.to_datetime(df_dynamics['Дата'])

# Строим линейный график выручки
plt.figure(figsize=(14, 6))
plt.plot(df_dynamics['Дата'], df_dynamics['Выручка за день (INR)'], color='orange', marker='o', linewidth=2)

plt.title('Динамика ежедневной выручки Amazon', fontsize=16, fontweight='bold')
plt.xlabel('Дата поездки / продажи', fontsize=12)
plt.ylabel('Выручка (INR)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
